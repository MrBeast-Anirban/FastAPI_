## Pydantic solves two problems
1. Type Validation
2. Data Validation


### 1. Data Validation

```python
# 1. Data validation -> user can enter data of any type (not ideal)
def insert_patient_data(name, age):
  print(name, age)

insert_patient_data("anirban", 26)
insert_patient_data("anirban", "twenty six")
```

```python
# Solution1 -> type hunting -> doesnot solves issue (no error if not of same type)
def insert_patient_data(name : str, age : int):
  print(name, age)

insert_patient_data("anirban", 26)
insert_patient_data("anirban", "twenty six")
```

```python
# Solution2 -> run time type checking -> not scalable
def insert_patient_data(name : str, age : int):
  if type(name) == str and type(age) == int:
    print(name, age)
  else:
    print("type error")

insert_patient_data("anirban", 26)
insert_patient_data("anirban", "twenty six")
```

```python
# ==========================( Suppress Type Coercion behaviour of Pydantic)==========================
# Its a possibility that pydantic will allow some value (type coercion) which should not be allowed
# Type coercion is the automatic or implicit conversion of a value from one data type to another

from pydantic import BaseModel, EmailStr, Field
from typing import List, Dict, Optional, Annotated

# step 1 :- create a pydantic class schema
class Patient(BaseModel):
  name : Annotated[str, Field(max_length = 50, title = 'Name of the Patient', description = 'Give the name of the patient in less than 50 characters.', examples = ['Anirban', 'Ankit'])]
  age : int
  email : EmailStr
  weight : Annotated[float, Field(gt = 0, lt = 120, strict = True)] # strict = True --> dont do type coercion
  married : Annotated[bool, Field(default = False, description = 'Is the patient married or not.')]
  allergies : Annotated[Optional[List[str]], Field(default = None, max_length = 5)]

# step 3 :- Pass pydantic object to function
def insert_patient_data(patient : Patient):
  print(patient.name)
  print(patient.age)
  print(patient.email)
  print(patient.weight)
  print(patient.married)
  print(patient.allergies)
  print("Inserted")

# step 2 :- Create pydantic object
patient_info = {'name' : 'anirban', 'age' : 26, 'email' : 'abc@gmail.com', 'weight' : '98.7', 'contact_details' : {'email' : 'abc@gmail.com', 'phone' : '134452352'}}
# error due to --> 'weight' : '98.7'
patient1 = Patient(**patient_info)

# Function calling with pydantic object patient1
insert_patient_data(patient1)
```

```python
# ==========================( Field Validator)==========================
# Field Validator helps in doing 1. Custom Validation over any variables/Fields 2. Transformation (ex. Name should start with Capital)
# Custom Validation -> The given emails should have hdfc.com or icici.com
# Transformation -> Patient's name should be in upper case

from pydantic import BaseModel, EmailStr, Field, field_validator
from typing import List, Dict, Optional, Annotated

# step 1 :- create a pydantic class schema
class Patient(BaseModel):
  name : str
  email : EmailStr
  age : int
  weight : float
  married : bool
  allergies : List[str]
  contact_details : Dict[str, str]

  # Custom Validation -> The given emails should have hdfc.com or icici.com
  @field_validator('email') # decorator
  @classmethod # is a built-in decorator used to define a method that is bound to the class itself rather than a specific object instance.
  def email_validator(cls, value):
    valid_domains = ['hdfc.com', 'icici.com']
    domain_name = value.split('@')[-1]
    if domain_name not in valid_domains:
      raise ValueError("Not a valid domain")
    return value

  # Transformation -> Patient's name should be in upper case
  @field_validator('name')
  @classmethod
  def transform_name(cls, value):
    return value.upper()




# step 3 :- Pass pydantic object to function
def insert_patient_data(patient : Patient):
  print(patient.name)
  print(patient.age)
  print(patient.email)
  print(patient.weight)
  print(patient.married)
  print(patient.allergies)
  print(patient.contact_details)
  print("Inserted")

# step 2 :- Create pydantic object
patient_info = {'name' : 'anirban', 'age' : 26, 'email' : 'abc@hdfc.com', 'weight' : '98.7', 'married' : True, 'allergies' : ['pollen', 'dust'], 'contact_details' : {'email' : 'abc@gmail.com', 'phone' : '134452352'}}
patient1 = Patient(**patient_info)

# Function calling with pydantic object patient1
insert_patient_data(patient1)
```

```python
# ==========================( Model Validator )==========================
# If the validation of a field is dependent on other fields
# Example : If age > 60 there should be a emergency contact detail

from pydantic import BaseModel, EmailStr, Field, field_validator, model_validator
from typing import List, Dict, Optional, Annotated

# step 1 :- create a pydantic class schema
class Patient(BaseModel):
  name : str
  email : EmailStr
  age : int
  weight : float
  married : bool
  allergies : List[str]
  contact_details : Dict[str, str]


@model_validator(mode = 'after') # no field name since we want to apply it throughout the schema
def validate_emergency_contact(self):
  if self.age > 60 and 'emergency' not in self.contact_details.keys:
    raise ValueError('Patients older than 60 must have an emergency contact.')
  return self




# step 3 :- Pass pydantic object to function
def insert_patient_data(patient : Patient):
  print(patient.name)
  print(patient.age)
  print(patient.email)
  print(patient.weight)
  print(patient.married)
  print(patient.allergies)
  print(patient.contact_details)
  print("Inserted")

# step 2 :- Create pydantic object
patient_info = {'name' : 'anirban', 'age' : 65, 'email' : 'abc@hdfc.com', 'weight' : '98.7', 'married' : True, 'allergies' : ['pollen', 'dust'], 'contact_details' : {'emergency': '3489349', 'email' : 'abc@gmail.com', 'phone' : '134452352'}}
patient1 = Patient(**patient_info)

# Function calling with pydantic object patient1
insert_patient_data(patient1)
```

```python
# ==========================( Computed Field )==========================
# The fields where user doesnot provide values. The values of that fields are set based on other fields
# Compute 'BMI' using 'height' and 'weight' field

from pydantic import BaseModel, EmailStr, Field, field_validator, model_validator, computed_field
from typing import List, Dict, Optional, Annotated

# step 1 :- create a pydantic class schema
class Patient(BaseModel):
  name : str
  email : EmailStr
  age : int
  weight : float #kgs
  height : float #metres
  married : bool
  allergies : List[str]
  contact_details : Dict[str, str]


  @computed_field
  @property
  def bmi(self) -> float:
    v = round(self.weight/(self.height**2), 2)
    return v



# step 3 :- Pass pydantic object to function
def insert_patient_data(patient : Patient):
  print(patient.name)
  print(patient.age)
  print(patient.email)
  print(patient.weight)
  print(patient.height)
  print("BMI", patient.bmi)
  print(patient.married)
  print(patient.allergies)
  print(patient.contact_details)
  print("Inserted")

# step 2 :- Create pydantic object
patient_info = {'name' : 'anirban', 'age' : 65, 'email' : 'abc@hdfc.com', 'weight' : '75.2', 'height': 1.72, 'married' : True, 'allergies' : ['pollen', 'dust'], 'contact_details' : {'emergency': '3489349', 'email' : 'abc@gmail.com', 'phone' : '134452352'}}
patient1 = Patient(**patient_info)

# Function calling with pydantic object patient1
insert_patient_data(patient1)
```

```python
# ==========================( Nested Model )==========================
from pydantic import BaseModel

class Address(BaseModel):
  city : str
  state : str
  pincode : str

class Patient(BaseModel):
  name : str
  age : int
  gender : str
  address : Address # 'house 2, sector 66, gurgaon, haryana, 122002'


address_dict = {'city' : 'gurgaon', 'state' : 'haryana', 'pincode' : '122002'}
address1 = Address(**address_dict)

patient_dict = {'name' : 'anirban', 'gender' : 'male', 'age' : 26, 'address' : address1}
patient1 = Patient(**patient_dict)

print(patient1)
print(patient1.name)
print(patient1.address.city)
print(patient1.address.pincode)




# Better organization of related data (e.g., vitals, address, insurance)

# Reusability: Use Vitals in multiple models (e.g., Patient, MedicalRecord)

# Readability: Easier for developers and API consumers to understand

# Validation: Nested models are validated automatically—no extra work needed
```

```python
# ==========================( Serialization : Export Pydantic Model Objects as Python Dictionary or Json )==========================
from pydantic import BaseModel

class Address(BaseModel):
  city : str
  state : str
  pincode : str

class Patient(BaseModel):
  name : str
  age : int
  gender : str
  address : Address # 'house 2, sector 66, gurgaon, haryana, 122002'


address_dict = {'city' : 'gurgaon', 'state' : 'haryana', 'pincode' : '122002'}
address1 = Address(**address_dict)

patient_dict = {'name' : 'anirban', 'gender' : 'male', 'age' : 26, 'address' : address1}
patient1 = Patient(**patient_dict)


# using model_dump() --> convert to dictionary
temp = patient1.model_dump()
print(temp)
print(type(temp))


# using model_dump() --> convert to json
temp2 = patient1.model_dump_json()
print(temp2)
print(type(temp2))


# model_dump(include = []) --> can control which fields to be exported
temp3 = patient1.model_dump(include = ['name', 'gender'])
print(temp3)
print(type(temp3))


# model_dump(exclude = []) --> can control which fields not to be exported
temp4 = patient1.model_dump(exclude = ['name', 'gender', {'address':['state']}])
print(temp4)
print(type(temp3))


# model.dump(exclude_unset = True) --> Whichever field is not set will not be exported
address_dict2 = {'city' : 'gurgaon', 'state' : 'haryana', 'pincode' : '122002'}
address2 = Address(**address_dict)

patient_dict2 = {'name' : 'anirban', 'age' : 26, 'address' : address1}
patient2 = Patient(**patient_dict)

temp3 = patient2.model_dump(exclude_unset = True)
print(temp3)
print(type(temp3))
```
